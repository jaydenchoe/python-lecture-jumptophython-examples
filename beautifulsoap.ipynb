## 📜 1단계: 웹사이트 정보 가져오기 (requests.get)

자, 이제 본격적으로 스크레이핑을 시작해 볼까요?

먼저, `requests` 라이브러리를 `import` (불러오기)해서 웹사이트에 접속해 볼게요. `get()` 함수에 우리가 정보를 가져오고 싶은 사이트의 주소(URL)를 넣어주면 돼요.

우리는 '가짜 파이썬 직업'이라는 연습용 사이트를 이용할 거예요. 이 사이트는 스크레이핑 연습을 위해 만들어진 고마운 사이트랍니다. 🙏

`URL = "https://realpython.github.io/fake-jobs/"`
`page = requests.get(URL)`

이 코드를 실행하면 `page`라는 변수에 웹사이트의 모든 응답 정보(HTML 코드, 접속 상태 등)가 저장됩니다.

````python
import requests

# 우리가 정보를 가져올 웹사이트 주소입니다.
URL = "https://realpython.github.io/fake-jobs/"

# requests.get()으로 웹사이트에 접속합니다.
page = requests.get(URL)

# page.status_code는 접속 상태를 알려줘요. 200이 나오면 "성공!"이라는 뜻입니다.
print(f"웹사이트 접속 상태: {page.status_code}")

# page.text에는 웹사이트의 전체 HTML 코드가 문자열로 저장되어 있어요.
# 너무 기니까 앞 500자만 잘라서 확인해 볼까요?
print("\n--- 웹사이트 HTML 코드 (앞 500자) ---")
print(page.text[:500])
````

## 🥣 2단계: HTML 코드 '요리하기' (BeautifulSoup)

방금 가져온 HTML 코드는 너무 복잡해서 우리가 원하는 정보를 찾기 어려워요. 😵

이때 'Beautiful Soup' (아름다운 수프)이 등장합니다! 이름처럼 복잡한 HTML(tag soup)을 멋지게 요리해서 우리가 다루기 쉽게 '객체'라는 형태로 만들어줘요.

`from bs4 import BeautifulSoup`
`soup = BeautifulSoup(page.content, "html.parser")`

`page.text` 대신 `page.content`를 쓰는 이유는, 간혹 한글 등이 깨지는 '인코딩' 문제를 `content`(bytes)가 더 잘 처리해주기 때문이에요. 'html.parser'는 파이썬의 기본 분석기입니다.

````python
from bs4 import BeautifulSoup

# 위에서 'page' 변수에 저장한 내용 중 .content를 BeautifulSoup에 넣어줍니다.
# 'html.parser'를 사용해서 HTML을 분석하라고 알려줍니다.
soup = BeautifulSoup(page.content, "html.parser")

# .prettify()를 쓰면 정리된 HTML 코드를 예쁘게 들여쓰기해서 보여줘요.
# 이것도 꽤 기니까, 앞 1000자만 잘라서 볼게요!
print("--- BeautifulSoup이 예쁘게 정리한 HTML (앞 1000자) ---")
print(soup.prettify()[:1000])
````

## 🎯 3단계: 원하는 정보 콕 집어 찾기 (find, id)

이제 '수프(soup)'가 준비됐으니, 원하는 재료(정보)를 찾아볼까요?

웹사이트를 브라우저에서 '개발자 도구'(F12 또는 마우스 우클릭 > 검사)로 살펴보면, **모든 구인 공고가 'ResultsContainer'라는 `id`를 가진 `<div>` 태그 안에** 들어있는 것을 알 수 있어요.

`id`는 사람의 주민등록번호처럼 웹페이지에서 **딱 하나만 존재하는 고유한 이름**이에요.

`soup.find()` 메서드를 사용하면 이 'ResultsContainer' 부분만 쏙 골라낼 수 있습니다.

````python
# soup (전체 HTML) 안에서 id가 'ResultsContainer'인 요소를 '찾아줘(find)'
results = soup.find(id="ResultsContainer")

# 'results'의 내용도 한번 예쁘게 출력해볼까요?
# 아까보다 내용이 훨씬 줄어들고, 구인 공고 부분만 남았을 거예요!
print("--- 'ResultsContainer' 부분의 HTML (앞 1000자) ---")
print(results.prettify()[:1000])
````

## 📚 4단계: 여러 개 정보 한 번에 가져오기 (find_all, class)

이제 'ResultsContainer' (`results` 변수) 안에서 각각의 '구인 공고 카드'를 모두 찾아야 해요.

개발자 도구로 보니, 각 공고 카드는 **'card-content'라는 `class`를 공통으로** 가지고 있네요.

`class`는 '반' 이름처럼 **여러 요소가 함께 가질 수 있는 그룹 이름**이에요.

`results` 안에서 `class`가 'card-content'인 **'모든(all)'** 요소를 찾으려면 `find_all()` 메서드를 씁니다.

**⚠️ 중요!**: 파이썬에서는 'class'가 이미 '클래스(class)를 만든다'는 특별한 의미로 사용되고 있어요. 그래서 HTML의 'class' 속성을 찾을 때는 `class_`라고 언더바(_)를 붙여서 구분해줍니다.

````python
# 'results' (공고 전체 묶음) 안에서,
# 'div' 태그이면서 class가 'card-content'인 *모든* 요소를 찾아 리스트로 만듭니다.
job_cards = results.find_all("div", class_="card-content")

# job_cards에는 몇 개의 공고 카드가 리스트 형태로 들어있을까요?
print(f"찾아낸 총 구인 공고 수: {len(job_cards)}개")

# 그 중 첫 번째(인덱스 0번) 공고 카드 내용만 한번 살펴볼게요.
print("\n--- 첫 번째 공고 카드 ---")
print(job_cards[0].prettify())
````

## 🔍 5단계: 카드에서 원하는 글자만 쏙! ( .text, .strip() )

이제 거의 다 왔어요! 🤩

각 '공고 카드'(`job_card`) 안에는 우리가 진짜 원하는 '제목', '회사명', '위치' 정보가 들어있어요.

개발자 도구로 다시 확인해보니, 이 정보들은 각각 'title', 'company', 'location'이라는 `class`를 가진 태그 안에 들어있네요.

`find()` 메서드로 각각을 찾은 다음, `.text` 속성을 이용하면 `<p>`, `<h2>` 같은 HTML 태그는 쏙 빼고 순수한 '글자'만 가져올 수 있어요.

글자 양옆에 불필요한 공백이나 줄바꿈이 있을 수 있으니, `.strip()` 메서드를 붙여서 깔끔하게 다듬어 줍니다. (예: `"  안녕하세요  \n"` -> `"안녕하세요"`)

````python
# job_cards 리스트에 있는 각 'job_card'를 하나씩 꺼내서 반복합니다.
print("--- 전체 구인 공고 목록 ---")

for job_card in job_cards:
  # 제목 찾기 (<h2> 태그, class='title')
  title_element = job_card.find("h2", class_="title")

  # 회사명 찾기 (<h3> 태그, class='company')
  company_element = job_card.find("h3", class_="company")

  # 위치 찾기 (<p> 태그, class='location')
  location_element = job_card.find("p", class_="location")

  # .text.strip()으로 글자만 깔끔하게 추출해서 출력
  print(f"제목: {title_element.text.strip()}")
  print(f"회사: {company_element.text.strip()}")
  print(f"위치: {location_element.text.strip()}")
  print("-" * 20) # 구분선
````

## 🔬 6단계: 내가 원하는 '파이썬' 공고만 필터링하기 (lambda)

모든 공고가 아니라, 'Python'이라는 단어가 포함된 공고만 보고 싶어요.

`find_all()` 메서드의 `string` 인자를 사용하면 텍스트 내용으로도 필터링할 수 있어요.

그런데 'Python', 'python', 'PYTHON' 등 대소문자가 다를 수 있겠죠? 이럴 땐 'lambda 함수'라는 멋진 기능을 사용합니다.

`lambda text: "python" in text.lower()`

이 코드는 "어떤 텍스트(text)가 들어오든, 그 텍스트를 전부 소문자(`lower()`)로 바꾼 다음 'python'이라는 단어가 포함(`in`)되어 있는지 확인해줘!"라는 뜻의 '미니 함수'예요.

````python
# 'results' (공고 전체 묶음) 안에서 <h2> 태그 중에...
# 텍스트를 소문자로 바꿨을 때 'python'이 포함된 모든 요소를 찾습니다.
python_job_titles = results.find_all(
    "h2",
    string=lambda text: "python" in text.lower()
)

# 몇 개나 찾았는지, 그리고 그 제목들은 무엇인지 출력해볼까요?
print(f"'Python' 키워드가 포함된 공고 수: {len(python_job_titles)}개")
print("\n--- 'Python' 관련 공고 제목 ---")

for title in python_job_titles:
  print(title.text.strip())
````

## 👨‍👩‍👧 7단계: 필터링한 공고의 '전체' 정보 가져오기 (부모 요소 찾기)

어라? 방금은 '제목(`<h2>`)'만 가져왔네요. '회사명'과 '위치' 정보도 같이 가져오고 싶어요.

하지만 `python_job_titles` 리스트에는 `<h2>` 요소만 들어있어서, `find("h3", class_="company")`를 실행하면 `None` (없음)이 반환되고 오류가 날 거예요. (AttributeError: 'NoneType' object has no attribute 'text')

이걸 어떻게 해결할까요?

**'제목' 태그는 '공고 카드'(`card-content`) 안에 속해있죠.** 즉, '공고 카드'는 '제목'의 '부모' 혹은 '조상' 뻘 되는 요소예요.

Beautiful Soup에서는 `.parent` 속성을 이용해 바로 위 부모 요소로 올라갈 수 있어요.
우리가 찾은 `<h2>` (제목) 태그에서 `.parent.parent.parent` 이렇게 3번 올라가면 'card-content' <div>를 다시 만날 수 있습니다! (이건 웹사이트 구조마다 다르니, 개발자 도구로 확인하는 습관이 중요해요!)

````python
# 위에서 찾은 'python_job_titles' (<h2> 요소들)을 순회합니다.
print("--- 'Python' 관련 공고 상세 정보 ---")

for h2_element in python_job_titles:
  # 부모의 부모의 부모 요소로 올라가 '공고 카드' 전체를 가져옵니다.
  job_card = h2_element.parent.parent.parent

  # 이제 5단계에서 했던 것과 똑같이 카드 안에서 회사명과 위치를 찾을 수 있어요.
  company_element = job_card.find("h3", class_="company")
  location_element = job_card.find("p", class_="location")

  print(f"제목: {h2_element.text.strip()}")
  print(f"회사: {company_element.text.strip()}")
  print(f"위치: {location_element.text.strip()}")
  print("-" * 20)
````

## 🔗 8단계: 숨겨진 '링크' 주소 가져오기 (속성 값 추출)

마지막으로, 'Apply' 버튼에 숨겨진 '지원하기' 링크(URL)를 가져와 볼게요.

링크는 `<a>` 태그로 만들어지고, 실제 주소는 `href`라는 '속성(attribute)' 안에 들어있어요.
(예: `<a href="https://google.com">구글로 가기</a>`)

`'Apply'`라는 글자는 `.text`로 가져올 수 있지만, `href` 속성값은 `.text`로 가져올 수 없어요.

이럴 땐 **`['href']`**처럼 딕셔너리에서 값을 꺼내듯이 속성 이름을 대괄호 안에 넣어주면 됩니다.

각 공고 카드에는 'Learn', 'Apply' 2개의 링크(`<a>` 태그)가 있어요. 우리가 필요한 건 두 번째(파이썬 리스트 인덱스로는 `[1]`) 링크입니다.

````python
# 다시 한번 '파이썬' 공고 카드들을 순회합니다.
print("--- 'Python' 관련 공고 지원 링크 ---")

for h2_element in python_job_titles:
  job_card = h2_element.parent.parent.parent

  # 이번엔 'Apply' 링크를 찾아봅니다.
  # 1. job_card 안의 모든 <a> 태그를 찾는다. (2개가 리스트로 찾아짐)
  all_links = job_card.find_all("a")

  # 2. 그 중에서 두 번째([1]) 링크를 선택한다.
  # 3. 선택한 링크의 'href' 속성 값을 가져온다.
  apply_link = all_links[1]["href"]

  print(f"제목: {h2_element.text.strip()}")
  print(f"지원 링크: {apply_link}")
  print("-" * 20)
````

## 🎉 축하합니다! 최종 코드 완성! 🎉

지금까지 배운 모든 조각을 하나로 합치면 아래와 같은 멋진 웹 스크레이퍼가 완성됩니다.

이 코드를 실행하면, '가짜 파이썬 직업' 사이트에서 'python' 키워드가 들어간 직업의 '제목', '회사명', '위치', '지원 링크'를 한 번에 쫙 뽑아줍니다.

여러분은 이제 웹 스크레이핑의 기본 원리를 마스터하셨습니다! 🥳

````python
import requests
from bs4 import BeautifulSoup

URL = "https://realpython.github.io/fake-jobs/"
page = requests.get(URL)

soup = BeautifulSoup(page.content, "html.parser")
results = soup.find(id="ResultsContainer")

# <h2> 태그이면서 텍스트에 'python'이 포함된 모든 요소를 찾기
python_job_titles = results.find_all(
    "h2", string=lambda text: "python" in text.lower()
)

print(f"총 {len(python_job_titles)}개의 '파이썬' 관련 공고를 찾았습니다.\n")
print("=" * 30)

# 찾은 <h2> 태그들을 순회하면서
for h2_element in python_job_titles:
  # 부모의 부모의 부모 요소(카드 전체)를 찾고
  job_card = h2_element.parent.parent.parent

  # 카드 안에서 회사명, 위치, 지원 링크를 각각 찾기
  company_element = job_card.find("h3", class_="company")
  location_element = job_card.find("p", class_="location")
  link_url = job_card.find_all("a")[1]["href"]

  # 깔끔하게 출력하기
  print(f"🐍 {h2_element.text.strip()}")
  print(f"🏢 {company_element.text.strip()}")
  print(f"📍 {location_element.text.strip()}")
  print(f"🔗 {link_url}\n")
  print("-" * 30)
````

````python
# -*- coding: utf-8 -*-
import requests
from bs4 import BeautifulSoup

url = "https://ko.wikipedia.org/wiki/대한민국의_영화_흥행_기록"
headers = {"User-Agent": "Mozilla/5.0"}

resp = requests.get(url, headers=headers, timeout=10)
resp.encoding = "utf-8"
soup = BeautifulSoup(resp.text, "html.parser")

# wikitable 클래스 표를 찾기
table = soup.find("table", class_="wikitable")
rows = table.find_all("tr")[1:11]  # 헤더 제외, 상위 10개 행

print("--- 한국 영화 흥행 TOP 10 ---")
for i, row in enumerate(rows, 1):
    cols = row.find_all("td")
    if len(cols) < 2:
        continue
    title = cols[1].get_text(strip=True)
    print(f"{i}위: {title}")
````

**문제 3:** 한국 영화 랭킹 1위 영화의 '링크(href)' 가져오기  
**힌트:** 1위 영화의 제목은 `<a>` 태그 안에 있으며, `['href']` 속성에 링크가 들어 있어요.  
위키백과 링크는 `/wiki/...` 형태이므로 앞에 `https://ko.wikipedia.org`를 붙여주세요.


````python
# -*- coding: utf-8 -*-
import requests
from bs4 import BeautifulSoup

url = "https://ko.wikipedia.org/wiki/대한민국의_영화_흥행_기록"
headers = {"User-Agent": "Mozilla/5.0"}

resp = requests.get(url, headers=headers, timeout=10)
resp.encoding = "utf-8"
soup = BeautifulSoup(resp.text, "html.parser")

# 첫 번째 wikitable을 찾고, 첫 번째 데이터 행의 두 번째 열에서 <a> 태그 추출
table = soup.find("table", class_="wikitable")
first_row = table.find_all("tr")[1]
cols = first_row.find_all("td")

a_tag = cols[1].find("a")
if a_tag and a_tag.has_attr("href"):
    href = a_tag["href"]
    if href.startswith("/"):
        href = "https://ko.wikipedia.org" + href
    print("1위 영화 링크:", href)
else:
    print("링크를 찾을 수 없습니다.")
````

**문제 9:** 알라딘 종합 베스트셀러 TOP 3 의 제목·저자·가격 가져오기  
**URL:** https://www.aladin.co.kr/shop/common/wbest.aspx?BranchType=1  
**힌트:** 제목은 a.bo3 태그, 저자는 같은 책 박스(div.ss_book_box) 안 li 태그 중 ‘지은이’ 포함 부분,  
가격은 span.ss_p2 strong 태그에서 찾아요.


````python
# -*- coding: utf-8 -*-
import requests
from bs4 import BeautifulSoup

url = "https://www.aladin.co.kr/shop/common/wbest.aspx?BranchType=1"
headers = {"User-Agent": "Mozilla/5.0"}

resp = requests.get(url, headers=headers, timeout=10)
resp.raise_for_status()
soup = BeautifulSoup(resp.text, "html.parser")

books = soup.select("div.ss_book_box")[:3]

print("--- 알라딘 베스트셀러 TOP 3 ---")

for i, book in enumerate(books, 1):
    # 제목
    title_tag = book.select_one("a.bo3")
    title = title_tag.get_text(strip=True) if title_tag else "제목 없음"

    # 저자 — '지은이' 문자열이 포함된 li를 느슨하게 탐색
    author = "저자 정보 없음"
    for li in book.select("li"):
        txt = li.get_text(strip=True)
        if "지은이" in txt or "(지은이)" in txt or "저자" in txt:
            author = txt
            break

    # 가격
    price_tag = book.select_one("span.ss_p2 strong")
    price = price_tag.get_text(strip=True) if price_tag else "가격 정보 없음"

    print(f"[{i}위] {title}")
    print(f"  {author}")
    print(f"  가격: {price}")
    print("-" * 30)
````

### 💡 왜 가격은 안 나올까?

알라딘·교보문고·YES24 같은 상업 사이트의 **가격 정보는 자주 바뀌기 때문**에,  
서버가 HTML을 보낼 때 숫자를 바로 넣지 않는다.  
대신 **브라우저에서 JavaScript가 실행될 때 동적으로 만들어지는 구조**다.

#### 📜 작동 원리
1. 사용자가 페이지를 열면 서버는 **상품의 기본 HTML 틀**만 보낸다.  
2. 브라우저 내부 JavaScript가 실행되면서  
   - 할인율 계산  
   - 회원가·쿠폰 적용  
   - 품절 표시  
   - 그 결과로 가격 텍스트를 DOM에 삽입한다.  
3. 따라서 `requests + BeautifulSoup`으로 가져오는 **원본 HTML에는 가격 숫자가 존재하지 않는다.**

#### ⚙️ 해결 방법
| 방법 | 설명 |
|------|------|
| **Selenium / Playwright** | 실제 브라우저를 띄워 JavaScript까지 실행한 뒤 완성된 HTML을 가져온다. |
| **공식 Open API (예: 알라딘 Open API)** | 인증키를 통해 JSON 형태로 가격 데이터를 직접 조회할 수 있다. |
| **정적 페이지 대체** | 위키백과, 해커뉴스 등 JS가 없는 정적 페이지를 연습용으로 활용한다. |

👉 **정리:**  
저자·제목은 서버 HTML에 이미 포함되어 있어서 `BeautifulSoup`으로 보이지만,  
가격은 JS로 렌더링되기 때문에 Colab에서 `requests`만으로는 볼 수 없다.

**문제 18:** 'Engineer'라는 단어가 포함된 모든 공고의 제목, 회사, 위치 가져오기
**힌트:** 6단계, 7단계에서 'python' 대신 'engineer'를 사용하세요.

````python
URL = "https://realpython.github.io/fake-jobs/"
page = requests.get(URL)
soup = BeautifulSoup(page.content, "html.parser")
results = soup.find(id="ResultsContainer")

# <h2> 태그이면서 텍스트에 'engineer'가 포함된 모든 요소를 찾기
engineer_job_titles = results.find_all(
    "h2", string=lambda text: "___" in text.lower()
)

print(f"--- 'Engineer' 관련 공고 ({len(engineer_job_titles)}개) ---")
for h2_element in engineer_job_titles:
  job_card = h2_element.parent.parent.parent
  company = job_card.find("h3", class_="company").text.strip()
  location = job_card.find("p", class_="location").text.strip()

  print(f"제목: {h2_element.text.strip()}")
  print(f"회사: {company}")
  print(f"위치: {location}")
  print("-" * 20)
````

**문제 19:** 'Stewartbury, AA' 지역의 모든 공고 제목, 회사명 가져오기
**힌트:** `find_all`로 `<p>` 태그이면서 `class`가 `location`인 것들 중, `.text.strip()`이 'Stewartbury, AA'와 같은 것을 찾으세요. (`string=` 인자 사용)

````python
URL = "https://realpython.github.io/fake-jobs/"
page = requests.get(URL)
soup = BeautifulSoup(page.content, "html.parser")
results = soup.find(id="ResultsContainer")

# <p> 태그, class 'location', 텍스트가 'Stewartbury, AA'인 모든 요소 찾기
locations = results.find_all("p", class_="location", string="___")

print(f"--- 'Stewartbury, AA' 지역 공고 ({len(locations)}개) ---")
for loc_element in locations:
  # <p> 태그에서 부모(div.content)의 부모(div.card-content)로 올라가기
  job_card = loc_element.parent.parent
  title = job_card.find("h2", class_="title").text.strip()
  company = job_card.find("h3", class_="company").text.strip()

  print(f"제목: {title}")
  print(f"회사: {company}")
  print("-" * 20)
````
