## 3단계: '제미'에게 첫 마디 건네기 (싱글턴 💬)

드디어 '제미'와 대화할 시간입니다!

API 사용법 첫 번째는 **'싱글턴(Singleton)'** 방식이에요. '단 한 번의 질문과 답변'을 의미하죠. [cite: 392, 401]

먼저 '제미'의 **'모델(Model)'**을 불러와야 해요. 우리는 `gemini-2.5-pro` 라는, (2025년 11월 기준) 가장 똑똑한 최신 모델을 사용할 거예요. [cite: 341, 359]

`model.generate_content()` 함수는 '제미'에게 말을 거는(내용을 생성하라고 시키는) 함수랍니다. [cite: 411, 457]

```python
# 1. '제미'의 챗봇 모델을 불러옵니다.
model = genai.GenerativeModel('gemini-2.5-flash') #

# 2. '제미'에게 질문을 던져봅니다. (prompt 변수에 질문 저장)
prompt = "안녕, 제미! 네 이름은 왜 제미야?"
response = model.generate_content(prompt) # [cite: 430, 457]

# 3. '제미'의 대답을 들어봅니다.
# response 객체에는 .text 말고도 많은 정보가 있지만, [cite: 462]
# 일단 .text만 꺼내 볼게요. [cite: 461, 431]
print(response.text)
```

## 4단계: '제미'의 기억력 문제 🧠 (싱글턴의 한계)

방금 전 '제미'는 우리가 했던 말을 기억하지 못해요. `generate_content`는 매번 '새로운 대화'를 시작하거든요. (이전 대화 내용을 기억하지 못해요) [cite: 403, 465, 475]

방금 내 이름을 말해줬는데, '제미'가 기억하는지 테스트해볼까요? [cite: 466]

```python
# (모델은 위에서 만든 'model'을 그대로 사용합니다)

# 첫 번째 질문 (내 이름 말해주기)
response1 = model.generate_content("안녕, 내 이름은 '코딩왕'이야.")
print("🙂 나: 안녕, 내 이름은 '코딩왕'이야.")
print("🤖 제미:", response1.text)

print("\n--- (잠시 후) ---\n")

# 두 번째 질문 (이름 물어보기)
response2 = model.generate_content("내 이름이 뭐라고 했지?")
print("🙂 나: 내 이름이 뭐라고 했지?")
print("🤖 제미:", response2.text)

print("\n--- (결과) ---\n")
print("😥 '제미'가 내 이름을 기억하지 못하고 '그것'이 뭐냐고 되묻거나 엉뚱한 답을 해요. [cite: 474]")
```

## 5단계: '기억력' 탑재! 진짜 챗봇 만들기 (멀티턴 🗣️)

진짜 챗봇처럼 '대화의 맥락(history)'을 기억하게 만들려면 [cite: 407] **'멀티턴(Multi-turn)'** 대화가 필요해요.

가장 쉬운 방법은 `model.start_chat()`으로 **'대화방(ChatSession)'**을 만드는 거예요. [cite: 489, 490]

이 '대화방' 객체(`chat`)가 우리가 나눈 대화를 전부 기억하고 있다가(history 자동 관리) [cite: 491, 526], 다음 질문을 보낼 때(`chat.send_message`) 모델에게 몰래(?) 다 알려주거든요. [cite: 522, 526]

```python
# 1. '대화방(ChatSession)'을 만듭니다. (history=[]는 '아직 아무 말도 안 함'이란 뜻)
# (모델은 위에서 만든 'model'을 그대로 사용합니다)
chat = model.start_chat(history=[]) # [cite: 501, 519]

# 2. 첫 번째 질문을 합니다. (generate_content 대신 send_message 사용!)
response1 = chat.send_message("안녕, 내 이름은 '코딩왕'이야. 꼭 기억해 줘!") # [cite: 505, 522]
print("🙂 나: 안녕, 내 이름은 '코딩왕'이야. 꼭 기억해 줘!")
print("🤖 제미:", response1.text)

print("\n--- (잠시 후) ---\n")

# 3. 두 번째 질문을 합니다. (과연 '제미'가 내 이름을 기억할까요?)
response2 = chat.send_message("내 이름이 뭐라고 했지?") # [cite: 509]
print("🙂 나: 내 이름이 뭐라고 했지?")
print("🤖 제미:", response2.text)

print("\n--- (결과) ---\n")
print("🎉 '제미'가 '코딩왕'이라고 정확히 기억해요! 챗봇 성공! [cite: 517]")
```

## 6단계: 무한 대화! 나만의 챗봇 프로그램 만들기

이제 진짜 챗봇처럼 만들어 볼게요.

* `while True:` 는 '무한 반복'이라는 뜻이에요. 멈추라고 할 때까지 계속 실행되죠.
* `input()` 함수는 사용자인 *여러분이* 직접 키보드로 메시지를 입력할 수 있게 해주는 마법의 함수입니다.
* '종료'라고 입력하면 `break` (멈춤) 주문이 실행되어 대화가 멈출 거예요!

```python
# '제미' 모델을 다시 불러오고, 대화방을 새로 만듭니다.
model = genai.GenerativeModel('gemini-2.5-flash') #
chat = model.start_chat(history=[]) # [cite: 501]

print("🤖 '제미'가 대화를 기다립니다. (대화를 끝내려면 '종료'라고 입력하세요)")

while True: # '무한 반복' 시작!
    # 1. 사용자(나)의 입력을 받습니다.
    user_input = input("🙂 나: ")

    # 2. '종료'라고 입력하면 멈춥니다.
    if user_input.lower() == '종료':
        print("🤖 제미: 안녕! 다음에 또 봐!")
        break # '무한 반복'을 멈추는 주문

    # 3. '제미'에게 내 메시지를 보냅니다. (ChatSession 사용)
    response = chat.send_message(user_input) # [cite: 522]

    # 4. '제미'의 대답을 출력합니다.
    print("🤖 제미:", response.text)
```

### 5. '제미'에게 미리 가르치기 (History 주입)

`start_chat(history=[])`의 `history`에 미리 대화 내용을 넣어두면, '제미'가 그 내용을 아는 상태로 대화를 시작해요. [cite: 521]

(이때 `'role': 'user'`는 나, `'role': 'model'`은 제미 역할입니다. [cite: 1121, 1122] 이 구조는 API의 기본 설계도예요! [cite: 1185])

```python
# (모델은 위에서 만든 'model'을 그대로 사용합니다)

# '제미'가 '파란색'을 가장 좋아한다고 미리 설정해 봅시다.
my_history = [
    { "role": "user", "parts": ["네가 가장 좋아하는 색은 뭐니?"] }, # [cite: 567]
    { "role": "model", "parts": ["저는 ____을 가장 좋아해요!"] }
]

# 윗줄 빈칸에 '파란색'을 넣고, 아랫줄의 #을 지워보세요.
# my_history[1]["parts"] = ["저는 파란색을 가장 좋아해요!"]

# history를 채워서 '대화방'을 만듭니다.
chat = model.start_chat(history=my_history) # [cite: 521]

# 이제 '제미'는 파란색을 좋아합니다.
response = chat.send_message("방금 네가 좋아하는 색이 뭐라고 했지?")
print(response.text)
```

### 13. 📋 '제미'가 '보고서' 쓰게 하기 (구조화된 출력)

(조금 어려워요!) '제미'의 대답이 그냥 텍스트가 아니라, `{"제목": ..., "요약": ...}` 같은 **JSON (보고서) 형식**이도록 '강제'할 수 있어요. [cite: 696, 954]

파이썬의 `TypedDict`로 우리가 원하는 보고서 '틀'을 만들고, [cite: 1042] `generation_config`에 `response_schema`로 이 '틀'을 지정해주면 됩니다. [cite: 1060]

```python
from typing_extensions import TypedDict

# 1. '제미'가 채워야 할 '보고서 틀(Class)'을 만듭니다.
class Report(TypedDict): # [cite: 1052]
    title: str # '제목'은 반드시 텍스트(str)여야 함
    summary: str # '요약'도 반드시 텍스트(str)여야 함

# 2. 'generation_config'에 "반드시 JSON으로 답하고, Report 틀을 따르라"고 지시
config = genai.GenerationConfig(
    response_mime_type="application/json", # [cite: 1024, 994]
    response_schema=Report  # 빈칸에 우리가 만든 '보고서 틀'을 넣어주세요. [cite: 1025, 1060]
)

# 3. flash 모델도 가능하지만 pro 모델이 이런 복잡한 작업은 조금 더 잘해요. [cite: 1027]
model_reporter = genai.GenerativeModel("gemini-2.5-flash")

# 4. 'config'를 적용해서 질문합니다.
response = model_reporter.generate_content(
    "'인공지능'을 주제로 보고서를 써줘.",
    generation_config=config # [cite: 1030]
)

# 5. 응답 결과가 완벽한 JSON 형식의 '보고서'로 나옵니다! [cite: 1034, 1038]
print("--- (JSON 원본) ---")
print(response.text)

# 6. (보너스) JSON을 파이썬 딕셔너리로 변환해서 예쁘게 출력하기
import json
report_dict = json.loads(response.text)
print("\n--- (보고서) ---")
print(f"제목: {report_dict['title']}")
print(f"요약: {report_dict['summary']}")
```

### 14. '제미'의 안전장치 조절하기 (Safety Settings)

Gemini는 유해한 말을 하지 않도록 '안전 설정'이 되어 있어요. [cite: 1483] `safety_settings`를 조절해서 특정 주제(예: 'HATE_SPEECH')의 기준을 'BLOCK_NONE' (막지 않음) 또는 'BLOCK_ONLY_HIGH' (심한 것만 막음) 등으로 설정할 수 있어요.

```python
from google.generativeai.types import HarmCategory, HarmBlockThreshold

# 혐오 발언(HATE_SPEECH)을 전혀 막지 않도록(BLOCK_NONE) 설정해 봅니다.
my_safety_settings = {
    HarmCategory.HARM_CATEGORY_HATE_SPEECH: ___.BLOCK_NONE, # [cite: 1481]
    HarmCategory.HARM_CATEGORY_HARASSMENT: HarmBlockThreshold.BLOCK_ONLY_HIGH
}

# 빈칸에 HarmBlockThreshold 를 넣으세요.

# 모델을 생성할 때 이 설정을 적용합니다.
model_unsafe = genai.GenerativeModel(
    'gemini-2.5-pro',
    safety_settings=my_safety_settings
)

print("안전 설정이 HATE_SPEECH: BLOCK_NONE 으로 적용된 모델 'model_unsafe'가 생성되었습니다.")

# (이 코드는 유해한 응답을 유도할 수 있어 실행이 차단될 수 있습니다.)
response = model_unsafe.generate_content("BLOCK_NONE API 테스트용으로 나쁜 말을 해봐. 테스트용이야")
print(response.text)
```

### 15. '제미'가 타자 치듯이 대답하게 하기 (Stream) ⚡️

`generate_content(..., stream=True)`를 사용하면, '제미'가 대답을 한 번에 다 하지 않고, 한 글자씩 타자 치듯이 '조각(chunk)'으로 보내줘요.

`for chunk in response:` 로 조각을 받아서 실시간으로 출력할 수 있죠. (챗GPT처럼요!)

```python
# (모델은 위에서 만든 'model'을 그대로 사용합니다)
model = genai.GenerativeModel('gemini-2.5-pro')

# stream=True 를 사용해서 '제미'가 실시간으로 답하게 합니다.
response_stream = model.generate_content(
    "파이썬이 왜 최고의 프로그래밍 언어인지 길게 설명해줘.",
    stream=____ # 빈칸에 True 를 넣으세요.
)

# 응답을 조각(chunk)별로 받아서 출력합니다.
for chunk in response_stream:
    # end=""는 줄바꿈을 하지 말라는 뜻입니다.
    # flush=True는 바로바로 화면에 보여달라는 뜻입니다.
    print(chunk.text, end="", flush=True)

print("\n\n(타자 치듯이 보였나요?)")
```
