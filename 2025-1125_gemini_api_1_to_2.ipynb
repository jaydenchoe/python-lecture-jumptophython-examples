## 2단계: '제미'를 깨우는 '비밀 열쇠' 등록하기 🔑

구글 AI Studio에서 받은 **API Key**가 필요합니다.

1.  [Google AI Studio (클릭)](https://aistudio.google.com/app/apikey) 에서 키를 생성하고 복사하세요.
2.  코랩 왼쪽 메뉴의 **'🔑' (열쇠 아이콘)**을 클릭하세요.
3.  **'새 비밀 추가'**를 누르고, 이름은 **`GOOGLE_API_KEY`**, 값에는 복사한 키를 붙여넣으세요.
4.  '노트북 액세스'를 켜주세요.

```python
from google import genai
from google.colab import userdata

try:
    # 1. 코랩의 비밀 상자에서 키를 꺼냅니다.
    API_KEY = userdata.get('GOOGLE_API_KEY')

    # 2. 최신 방식: '클라이언트(Client)'를 만듭니다. (이게 제미의 본체예요!)
    client = genai.Client(api_key=API_KEY)

    print("✨ '제미'가 깨어날 준비가 되었어요! (최신 Gen AI SDK 연결 성공)")
except Exception as e:
    print(f"🔑 오류! 비밀 키를 가져오지 못했어요. {e}")
    print("왼쪽 🔑 아이콘에 'GOOGLE_API_KEY'를 저장했는지 확인해주세요!")
```

## 3단계: '제미'에게 첫 마디 건네기 (싱글턴 💬)

이제 `client.models.generate_content` 명령어로 대화를 시작해 봅시다.
우리는 빠르고 똑똑한 **`gemini-2.5-flash`** 모델을 사용할 거예요.

```python
# 1. '제미'에게 질문을 던져봅니다.
prompt = "안녕, 제미! 네 이름은 왜 제미야?"

# 2. 최신 SDK 사용법: client.models.generate_content
response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=prompt
)

# 3. 대답 출력
print(response.text)
```

## 4단계: '제미'의 기억력 문제 🧠 (싱글턴의 한계)

방금 전 방식은 '한 번 묻고 답하기'라서, 이전 대화를 기억하지 못해요.

```python
# 첫 번째 질문 (내 이름 말해주기)
response1 = client.models.generate_content(
    model='gemini-2.5-flash',
    contents="안녕, 내 이름은 '코딩왕'이야."
)
print("🙂 나: 안녕, 내 이름은 '코딩왕'이야.")
print("🤖 제미:", response1.text)

print("\n--- (잠시 후) ---\n")

# 두 번째 질문 (이름 물어보기)
response2 = client.models.generate_content(
    model='gemini-2.5-flash',
    contents="내 이름이 뭐라고 했지?"
)
print("🙂 나: 내 이름이 뭐라고 했지?")
print("🤖 제미:", response2.text)
print("\n(😥 기억을 못하죠?)")
```

## 5단계: '기억력' 탑재! 진짜 챗봇 만들기 (멀티턴 🗣️)

대화를 기억하려면 **`client.chats.create`** 기능으로 '대화방'을 만들어야 합니다.

```python
# 1. '대화방(Chat)'을 만듭니다. (history=[]는 빈 방으로 시작)
chat = client.chats.create(
    model='gemini-2.5-flash',
    history=[]
)

# 2. 첫 번째 질문 (send_message 사용!)
response1 = chat.send_message("안녕, 내 이름은 '코딩왕'이야. 꼭 기억해 줘!")
print("🙂 나: 안녕, 내 이름은 '코딩왕'이야. 꼭 기억해 줘!")
print("🤖 제미:", response1.text)

print("\n--- (잠시 후) ---\n")

# 3. 두 번째 질문
response2 = chat.send_message("내 이름이 뭐라고 했지?")
print("🙂 나: 내 이름이 뭐라고 했지?")
print("🤖 제미:", response2.text)
print("\n(🎉 이제 기억하네요!)")
```

## 6단계: 무한 대화! 나만의 챗봇 프로그램 만들기

```python
# 새 대화방 생성
chat = client.chats.create(model='gemini-2.5-flash')

print("🤖 '제미'가 대화를 기다립니다. (대화를 끝내려면 '종료'라고 입력하세요)")

while True:
    # 1. 사용자 입력 받기
    user_input = input("🙂 나: ")

    # 2. 종료 조건
    if user_input.lower() == '종료':
        print("🤖 제미: 안녕! 다음에 또 봐!")
        break

    # 3. 메시지 보내기
    response = chat.send_message(user_input)
    print("🤖 제미:", response.text)
```

### 7. ⚡️ 타자 치듯이 대답하기 (Streaming)

```python
# generate_content_stream 메서드를 사용합니다.
response_stream = client.models.generate_content_stream(
    model='gemini-2.5-pro',
    contents="파이썬의 장점을 3가지로 길게 설명해줘."
)

print("🤖 제미 (스트리밍 중): ")
for chunk in response_stream:
    # 실시간으로 글자가 찍힙니다.
    print(chunk.text, end="", flush=True)

print("\n\n(완료!)")
```
