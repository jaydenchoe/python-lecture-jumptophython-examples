## Part 1. 인터넷은 어떻게 작동할까?

코드를 짜기 전에, 우리가 인터넷과 어떻게 대화하는지 아주 쉬운 비유를 통해 알아봅시다.

### 📖 비유로 이해하는 인터넷: 거대한 글로벌 도서관

인터넷을 **전 세계 모든 정보(책)가 보관된, 24시간 열려있는 거대한 도서관**이라고 상상해 보세요.

* **웹사이트 (e.g., `google.com`)**: 도서관의 특정 코너나 서가 (`'과학 코너'`).
* **웹 페이지**: 서가에 꽂혀있는 한 권의 책 (`'코스모스'`).
* **URL (e.g., `https://www.google.com`)**: 이 거대한 도서관에서 책의 위치를 정확하게 알려주는 **'도서 청구 기호'**.

---

이 도서관에는 항상 두 명의 주인공이 등장합니다.

1.  **클라이언트 (Client, 손님)**
    * 바로 **'우리 자신'** 입니다. (정확히는 우리의 웹 브라우저나 우리가 만들 파이썬 프로그램)
    * 도서관에 찾아가 "이 책 좀 보여주세요"라고 **요청(Request)**하는 손님입니다.

2.  **서버 (Server, 사서)**
    * **'도서관 사서'** 입니다.
    * 손님의 요청을 받고, 거대한 책장(데이터베이스)에서 요청한 책(데이터)을 찾아서 건네주는 **응답(Response)**을 합니다.

> **우리가 웹 서핑을 하는 모든 순간은, 손님(클라이언트)이 사서(서버)에게 끊임없이 책을 요청하고 받는 과정의 연속입니다.**

---

### 🗣️ 그들의 대화 규칙: HTTP

전 세계 수많은 손님과 사서들이 뒤죽박죽 대화하면 혼란스럽겠죠? 그래서 모두가 따르는 공통된 대화 규칙, 즉 **프로토콜(Protocol)**이 필요합니다.

인터넷 세상에서 클라이언트와 서버가 사용하는 대화 규칙이 바로 **HTTP (HyperText Transfer Protocol)** 입니다.

* **HTTP**: "인터넷 도서관에서 정보를 주고받기 위한 공식적인 약속(언어)"
* **HTTPS**: HTTP + **Secure**. 손님과 사서가 둘만 아는 **암호**로 속삭이며 대화하는 것과 같습니다. 로그인, 결제 등 중요한 정보는 모두 HTTPS로 보호됩니다.

### 📥 요청의 종류: GET vs POST

손님은 사서에게 그냥 "책 주세요"라고 말하지 않고, 구체적인 의도를 전달해야 합니다. 이걸 **HTTP 메서드**라고 불러요. 가장 중요한 두 가지만 기억하세요!

#### 1. GET: "보여주세요!"
* 서버로부터 정보를 **'가져오기(Get)'** 위해 사용합니다.
* 서버의 데이터를 변경하지 않고, 오직 **읽기(Read)**만 할 때 사용됩니다.
* **예시**: 웹툰 목록 보기, 뉴스 기사 클릭하기, 상품 상세 페이지 보기.

#### 2. POST: "처리해주세요!"
* 클라이언트가 서버에 정보를 **'제출하기(Post)'** 위해 사용합니다.
* 서버에 새로운 정보를 **생성(Create)**하거나, 기존 정보를 **수정(Update)**할 때 주로 사용됩니다.
* **예시**: 로그인(ID/PW 제출), 회원가입(내 정보 제출), 게시글 작성(제목/내용 제출).


### ✅ 응답의 결과: 상태 코드 (Status Code)

사서(서버)는 손님(클라이언트)의 요청을 처리한 뒤, 그 결과가 어땠는지 숫자로 된 **'상태 코드'**를 가장 먼저 알려줍니다. 이 코드를 통해 우리는 요청이 성공했는지, 실패했다면 왜 실패했는지 알 수 있습니다.

* `2xx (Success)`: **성공!** 요청이 성공적으로 처리되었습니다.
    * `200 OK`: 가장 일반적인 성공. "요청하신 데이터 여기 있습니다."
* `3xx (Redirection)`: **다른 곳으로 안내!** "찾으시는 책은 저쪽 코너로 옮겨졌어요."
* `4xx (Client Error)`: **손님, 요청이 잘못됐어요!**
    * `404 Not Found`: "죄송하지만, 요청하신 책(페이지)은 우리 도서관에 없네요."
    * `401 Unauthorized`: "로그인 먼저 해주세요."
    * `403 Forbidden`: "로그인은 하셨지만, 이 자료는 관리자만 볼 수 있습니다."
* `5xx (Server Error)`: **죄송합니다, 저희 도서관에 문제가 생겼어요!**
    * `500 Internal Server Error`: "도서관 전산 시스템에 문제가 생겨서 책을 찾아드릴 수가 없어요."

### 2. 첫 번째 요청: `requests.get()`

웹 페이지의 내용을 가져오는 `GET` 요청은, 마법처럼 단 한 줄이면 충분합니다. Github의 API 서버에 간단한 요청을 보내봅시다.

````python
import requests

# requests.get() 함수에 URL(도서 청구 기호)을 전달하면 끝!
response = requests.get("https://api.github.com")

# 서버의 응답을 출력해봅시다.
print(response)
````

### 3. 응답 상자 열어보기: `Response` 객체

방금 받은 `response` 변수는 단순한 텍스트가 아니라, 서버가 보낸 응답의 모든 정보가 담겨있는 **'정보 상자(Response 객체)'**입니다.

이 상자에서 다양한 정보를 점(`.`)을 찍어 꺼내 볼 수 있습니다.

````python
# 가장 먼저 확인할 것: 상태 코드
print(f"상태 코드: {response.status_code}")

# '파이썬스러운' 상태 확인 방법: Response 객체는 성공(2xx)이면 True, 실패(4xx, 5xx)면 False로 평가됩니다.
if response:
    print("요청 성공!")
else:
    print("요청 실패!")
````

#### 자동으로 오류 발생시키기: `.raise_for_status()`

매번 `if`문으로 상태 코드를 확인하는 것이 번거로울 수 있습니다. `.raise_for_status()` 메서드는 응답 코드가 4xx 또는 5xx(오류)일 경우에만 **프로그램을 멈추고 에러를 발생**시킵니다. 성공적인 코드일 경우에는 아무 일도 일어나지 않습니다.

안정적인 프로그램을 위해 `try...except` 구문과 함께 사용하는 것이 좋습니다.

````python
try:
    # 일부러 없는 주소로 요청을 보내 404 에러를 유도해봅시다.
    response = requests.get("https://api.github.com/invalid-url")
    response.raise_for_status() # 여기서 404 에러가 발생하여 except 블록으로 점프!
except requests.exceptions.HTTPError as err:
    print(f"HTTP 오류가 발생했습니다: {err}")
````

### 4. 응답의 핵심: 내용물(Content) 보기

서버가 보내준 웹 페이지의 실제 내용은 크게 두 가지 속성으로 접근할 수 있습니다. 이 둘의 차이를 이해하는 것은 매우 중요합니다!

* `.text` (번역된 이야기책): 서버가 보낸 원본 데이터를 `requests`가 **문자열(str)**로 자동 번역해준 결과물입니다. (HTML, JSON 등 텍스트 기반 데이터에 사용)
* `.content` (원본 그대로의 데이터): 어떠한 변환도 거치지 않은 원본 **바이트(bytes)** 데이터입니다. (이미지, 동영상, PDF 등 텍스트가 아닌 데이터에 사용)

````python
response = requests.get("https://www.google.com")

# .text 속성: HTML 코드가 문자열(str) 형태로 출력됩니다.
print(f".text 타입: {type(response.text)}")
print(response.text[:200]) # 너무 기니까 앞 200글자만 출력

print("\n" + "="*50 + "\n")

# .content 속성: 원본 바이트(bytes) 데이터가 출력됩니다. 앞에 b' 가 붙어있어요.
print(f".content 타입: {type(response.content)}")
print(response.content[:200]) # 앞 200바이트만 출력
````

#### 마법의 메서드: `.json()`

최근 웹 서비스들은 **JSON**이라는 형식으로 데이터를 주고받는 것을 선호합니다. JSON은 파이썬의 딕셔너리(`{}`)와 리스트(`[]`) 구조와 거의 동일하여 다루기 매우 쉽습니다.

서버가 보낸 응답이 JSON 형식일 경우, `.json()` 메서드를 호출하면 마법처럼 **파이썬 딕셔너리나 리스트로 변환**해줍니다.

````python
response = requests.get("https://api.github.com")

# .json() 메서드를 호출하면...
data = response.json() # 마법처럼 파이썬 딕셔너리로 변신!

print(f"data의 타입: {type(data)}")

# 이제 딕셔너리처럼 자유롭게 키를 이용해 값을 꺼낼 수 있습니다.
print(f"current_user_url의 값: {data['current_user_url']}")
````

## Part 3. 더 구체적인 요청 만들기

단순히 페이지만 가져오는 것을 넘어, 우리가 원하는 조건을 달아서 요청하는 방법을 배워봅시다.

### 1. GET 요청에 날개 달기: 쿼리 스트링 (Query Strings)

URL 뒤에 `?`를 붙여 `key=value` 형태의 추가 정보를 전달하는 것을 **쿼리 스트링**이라고 합니다. "파이썬에 관한 책만 찾아주세요" 처럼 더 구체적인 조건을 달아 요청할 때 사용합니다.

예시: `https://www.google.com/search?q=python`
(`q`는 '검색어'를 의미하는 파라미터, `python`은 그 값입니다)

`requests`에서는 `params` 파라미터에 딕셔너리를 전달하는 것만으로 이 모든 것을 자동으로 처리해줍니다.

````python
# 우리가 원하는 파라미터를 딕셔너리로 깔끔하게 준비
payload = {'q': 'python', 'sort': 'stars'}

# requests.get()의 params 인자에 전달
response = requests.get("https://api.github.com/search/repositories", params=payload)

# requests가 자동으로 만든 최종 URL을 확인해 봅시다.
print(f"요청된 최종 URL: {response.url}")
````

### 2. 요청의 신분증: 요청 헤더 (Request Headers)

많은 웹사이트들은 자동화된 프로그램(봇)의 접근을 막기 위해 요청의 **'신분증'**을 확인합니다. 이 신분증이 바로 **`User-Agent`** 라는 헤더 값입니다.

`requests`의 기본 `User-Agent`는 'python-requests/버전'이라서 "나 파이썬 프로그램이오!"라고 광고하는 셈이라, 서버가 접근을 차단할 수 있습니다.

따라서 `User-Agent`를 **일반 웹 브라우저인 것처럼 바꿔서** 보내주면, 서버가 우리를 평범한 사용자로 인식하고 정상적인 페이지를 보여줄 확률이 높아집니다. (웹 크롤링 시 거의 필수!)

````python
# 일반적인 크롬 브라우저의 User-Agent 값 (그대로 복사해서 쓰면 됩니다)
my_headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

# headers 파라미터에 우리가 만든 딕셔너리를 전달합니다.
response = requests.get("https://www.naver.com", headers=my_headers)

# 이제 네이버 서버는 우리가 크롬 브라우저에서 접속한 것으로 인식합니다.
print(f"네이버 요청 상태 코드: {response.status_code}")
````

## Part 4. 데이터를 보내는 기술: POST 요청

`GET`이 서버의 정보를 '읽는' 행위였다면, `POST`는 서버에 정보를 '쓰는' 행위입니다. 로그인, 회원가입, 글쓰기 등에 사용되죠. `requests.post()` 함수를 사용하며, 데이터를 보내는 방식은 크게 두 가지입니다.

### 1. Form 데이터 보내기 (`data` 파라미터)

전통적인 웹사이트의 `<form>` 태그가 데이터를 보내는 방식입니다. `requests`가 딕셔너리를 `key1=value1&key2=value2` 형태의 문자열로 변환하여 전송합니다.

우리가 보낸 요청이 서버에 어떻게 도착했는지 그대로 다시 보여주는 '메아리' 사이트인 `httpbin.org`를 이용해 테스트해봅시다.

````python
URL = 'https://httpbin.org/post'
payload = {'username': 'alex', 'password': '123'}

# data 파라미터에 딕셔너리를 전달
response = requests.post(URL, data=payload)

# httpbin이 받은 데이터를 확인해봅시다.
# 'form'이라는 키 안에 우리가 보낸 데이터가 들어있습니다.
print(response.json()['form'])
````

### 2. JSON 데이터 보내기 (`json` 파라미터)

현대적인 API에서 주로 사용하는 방식입니다. `json` 파라미터를 사용하면 `requests`가 알아서 파이썬 딕셔너리를 JSON 형식의 문자열로 변환하고, `Content-Type` 헤더를 `application/json`으로 설정까지 해줍니다. **훨씬 편리합니다.**

````python
URL = 'https://httpbin.org/post'
payload = {'item_name': 'python_book', 'quantity': 1}

# json 파라미터에 딕셔너리를 전달
response = requests.post(URL, json=payload)

# httpbin이 받은 데이터를 확인해봅시다.
# 'json'이라는 키 안에 우리가 보낸 데이터가 들어있습니다.
print(response.json()['json'])
````

## Part 5. 프로 개발자의 `requests` 활용법

안정적이고 효율적인 프로그램을 만들기 위한 고급 기술들을 맛봅시다.

### 1. 무한정 기다리지 않기: 타임아웃 (Timeout)

특정 서버가 응답이 없을 때, 우리 프로그램이 무한정 기다리게 할 수는 없습니다. `timeout` 파라미터에 초(seconds) 단위로 시간을 지정하면, 해당 시간 안에 응답이 오지 않을 경우 `Timeout` 예외를 발생시킵니다.

**실무 팁**: 외부 API를 호출하는 모든 `requests` 코드에는 반드시 `timeout`을 설정하는 습관을 들이세요!

````python
from requests.exceptions import Timeout

try:
    # 일부러 응답이 없는 가상 주소로 요청을 보내고, 타임아웃을 3초로 설정합니다.
    # (실제로는 연결 자체가 안되어 ConnectionError가 먼저 발생할 수 있습니다)
    response = requests.get('http://10.255.255.1', timeout=3)
except Timeout:
    print("요청 시간이 초과되었습니다. 서버가 너무 느리거나 문제가 있는 것 같습니다.")
except requests.exceptions.ConnectionError as e:
    print(f"연결에 실패했습니다: {e}")
````

### 2. 상태를 기억하는 비서: 세션 (Session) 객체

한 사이트에서 로그인을 한 뒤, 여러 페이지에 걸쳐 로그인 상태를 유지하고 싶을 때 사용합니다. `requests.get()`을 매번 호출하는 것은 매번 다른 손님이 도서관에 찾아가는 것과 같지만, **`Session` 객체**를 사용하면 한 명의 손님이 계속해서 요청을 보내는 것과 같습니다.

**장점:**
* **쿠키(Cookie) 자동 관리**: 서버가 발급한 '로그인 출입증(쿠키)'을 자동으로 저장했다가 다음 요청부터 계속 제시합니다.
* **기본 설정 유지**: 헤더, 인증 정보 등을 한 번만 설정해두면 세션을 통해 보내는 모든 요청에 자동 적용됩니다.
* **성능 향상**: 서버와의 연결을 재사용하여 여러 번 요청을 보낼 때 훨씬 더 빠릅니다.

````python
# 1. with 구문으로 Session 객체를 생성합니다. (자동으로 리소스 정리)
with requests.Session() as s:
    # 2. 첫 번째 요청: 서버에게 'my_cookie'라는 쿠키를 '12345' 값으로 설정해달라고 요청
    s.get('https://httpbin.org/cookies/set/my_cookie/12345')

    # 3. 두 번째 요청: 같은 세션으로 다른 페이지에 접속
    # 세션은 방금 받은 쿠키('my_cookie=12345')를 자동으로 포함해서 요청을 보냅니다.
    response = s.get('https://httpbin.org/cookies')

    # 서버가 받은 쿠키를 그대로 돌려주는 것을 확인
    print(response.json())
````

### 실습 1. 오늘의 명언 제조기 (GET & JSON)

무료 명언 API에 접속해서, 실행할 때마다 새로운 영어 명언과 작가 정보를 가져와 출력하는 프로그램을 만듭니다.

````python
import requests

# 명언 API 주소
API_URL = "https://zenquotes.io/api/random"

print("오늘의 명언을 가져오는 중...\n")

try:
    # 1. API에 GET 요청을 보냅니다 (타임아웃 설정은 좋은 습관!)
    response = requests.get(API_URL, timeout=5)
    # 2. 요청이 실패했다면 에러를 발생시킵니다.
    response.raise_for_status()

    # 3. JSON 데이터를 파이썬 딕셔너리로 변환합니다.
    data = response.json()[0]

    # 4. 딕셔너리에서 명언 내용(content)과 작가(author)를 추출합니다.
    quote = data['q']
    author = data['a']

    # 5. 화면에 예쁘게 출력합니다.
    print(f'"{quote}"')
    print(f'- {author} -')

except requests.exceptions.RequestException as e:
    # 타임아웃, 연결 오류, HTTP 오류 등 모든 requests 관련 에러를 잡습니다.
    print(f"명언을 가져오는 데 실패했습니다: {e}")
````

### 실습 2. '나만의 포켓몬 도감' 만들기 (동적 GET & 데이터 파싱)

사용자로부터 포켓몬의 이름을 입력받아, PokéAPI에 접속해 해당 포켓몬의 상세 정보(이름, 타입, 키, 몸무게)를 가져와 보여주는 미니 도감을 만듭니다.

````python
import requests

# 1. 사용자로부터 조회하고 싶은 포켓몬의 이름을 영어로 입력받습니다.
pokemon_name = input("정보를 조회할 포켓몬의 이름을 영어로 입력하세요 (e.g., pikachu): ")

# 2. 사용자의 입력을 소문자로 바꾸고, f-string을 이용해 최종 API URL을 조립합니다.
API_URL = f"https://pokeapi.co/api/v2/pokemon/{pokemon_name.lower()}"

try:
    # 3. 조립된 URL로 GET 요청을 보냅니다.
    response = requests.get(API_URL, timeout=3)
    response.raise_for_status() # 포켓몬이 없으면 404 에러가 발생하며 except로 넘어갑니다.

    # 4. JSON 데이터를 딕셔너리로 변환합니다.
    data = response.json()

    # 5. 딕셔너리의 여러 계층을 탐색하여 원하는 정보를 추출합니다.
    print("\n" + "="*20)
    print(f"도감번호: #{data['id']}")
    print(f"이름: {data['name'].capitalize()}") # 첫 글자만 대문자로
    print(f"키: {data['height'] * 10} cm") # API 문서에 따르면 단위가 데시미터(dm)
    print(f"몸무게: {data['weight'] / 10} kg") # API 문서에 따르면 단위가 헥토그램(hg)

    # 6. 타입 정보는 리스트 안에 딕셔너리가 있는 중첩 구조입니다.
    print("타입:")
    types = [t['type']['name'] for t in data['types']]
    for t in types:
        print(f"- {t.capitalize()}")
    print("="*20)

# 7. raise_for_status()가 4xx, 5xx 에러를 발생시켰을 때 잡는 부분입니다.
except requests.exceptions.HTTPError:
    print(f"'\"{pokemon_name}\"' 포켓몬을 찾을 수 없습니다. 이름을 다시 확인해주세요.")
except requests.exceptions.RequestException as e:
    print(f"오류가 발생했습니다: {e}")
````

### 실습 3. 웹툰 최신화 알림 봇 (크롤링 첫걸음)

내가 좋아하는 웹툰 페이지에 접속해서, 페이지의 전체 HTML 소스 코드를 가져오는 프로그램을 만듭니다. (이것이 모든 웹 크롤링의 시작입니다!) 봇으로 오인받지 않도록 `User-Agent` 헤더를 설정하는 것이 핵심입니다.

````python
import requests

# 네이버 웹툰의 '화산귀환' 작품 페이지 주소를 예시로 사용합니다.
WEBTOON_URL = "https://comic.naver.com/webtoon/list?titleId=769209"

# 1. 서버를 속이기 위한 '위장 신분증(User-Agent)'을 만듭니다.
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

try:
    # 2. 요청을 보낼 때, headers 파라미터에 위에서 만든 '신분증'을 포함시킵니다.
    response = requests.get(WEBTOON_URL, headers=headers, timeout=3)
    response.raise_for_status()

    # 3. 웹 페이지의 전체 HTML 코드를 .text 속성으로 가져옵니다.
    html_code = response.text

    print(f"{WEBTOON_URL} 페이지의 HTML 코드를 성공적으로 가져왔습니다!")
    print("코드 길이:", len(html_code))
    print("--- HTML 코드 미리보기 (앞 500자) ---")
    print(html_code[:500])

except requests.exceptions.RequestException as e:
    print(f"페이지를 가져오는 데 실패했습니다: {e}")
````

### 실습 4. 파일 다운로드하기 (이미지 저장)

인터넷에 있는 이미지의 URL을 이용해, 해당 이미지를 내 컴퓨터(코랩 환경)에 파일로 저장합니다. 텍스트가 아닌 바이너리(binary) 데이터를 다룰 때는 `.content` 속성을 사용해야 합니다.

````python
import requests

# Real Python 로고 이미지 URL
IMAGE_URL = 'https://www.python.org/static/community_logos/python-logo.png'
# 저장할 파일 이름
FILENAME = 'python-logo.png'

try:
    print(f"'{FILENAME}' 다운로드를 시작합니다...")
    response = requests.get(IMAGE_URL, timeout=5)
    response.raise_for_status()

    # 1. 파일을 'wb'(write binary) 모드로 엽니다. 'b'가 핵심입니다!
    with open(FILENAME, 'wb') as f:
        # 2. 응답의 .content (원본 바이트 데이터)를 파일에 씁니다.
        f.write(response.content)

    print(f"'{FILENAME}' 파일이 성공적으로 저장되었습니다.")
    print("왼쪽 파일 탐색기 탭(폴더 아이콘)에서 확인해보세요!")

except requests.exceptions.RequestException as e:
    print(f"이미지 다운로드에 실패했습니다: {e}")
````

#### (심화) 대용량 파일 스트리밍 다운로드

만약 수 기가바이트(GB)에 달하는 동영상 파일을 다운로드한다면, `response.content`는 모든 내용을 메모리에 한꺼번에 올리려다 메모리 부족으로 프로그램이 멈출 수 있습니다. 이럴 땐 `stream=True` 옵션과 `.iter_content()` 메서드를 사용해 데이터를 '조각(chunk)' 단위로 나누어 처리해야 합니다.

````python
import requests

IMAGE_URL = 'https://www.python.org/static/community_logos/python-logo.png'
FILENAME = 'python-logo-streamed.png'

try:
    # 1. stream=True 옵션을 주어 응답 내용을 바로 메모리에 올리지 않도록 합니다.
    with requests.get(IMAGE_URL, stream=True, timeout=5) as r:
        r.raise_for_status()

        # 2. 파일을 'wb'(write binary) 모드로 엽니다.
        with open(FILENAME, 'wb') as f:
            print(f"'{FILENAME}' 스트리밍 다운로드를 시작합니다...")
            # 3. 응답 내용을 8192바이트(8KB)씩 나누어 순회합니다.
            for chunk in r.iter_content(chunk_size=8192):
                # 4. 나누어진 조각(chunk)을 파일에 씁니다.
                f.write(chunk)
            print(f"'{FILENAME}' 파일이 성공적으로 저장되었습니다.")

except requests.exceptions.RequestException as e:
    print(f"이미지 다운로드에 실패했습니다: {e}")
````

### 실습 5. 파일 업로드하기 (POST)

내 컴퓨터(코랩 환경)에 있는 파일을 `POST` 요청을 통해 서버에 업로드합니다. `files` 파라미터를 사용하는 방법을 배웁니다.

````python
import requests

URL = 'https://httpbin.org/post'

# 1. 업로드할 테스트 파일을 하나 만듭니다.
with open('my_report.txt', 'w', encoding='utf-8') as f:
    f.write('이것은 나의 보고서입니다.\n')
    f.write('파이썬은 정말 멋져요!\n')

try:
    # 2. 업로드할 파일을 'rb'(read binary) 모드로 엽니다.
    with open('my_report.txt', 'rb') as f:
        # 3. 'files' 파라미터에 딕셔너리 형태로 파일을 전달합니다.
        # - key ('report_file')는 서버에서 파일을 식별할 이름입니다.
        # - value (f)는 열려있는 파일 객체입니다.
        files = {'report_file': f}

        # 4. POST 요청으로 파일을 업로드합니다.
        response = requests.post(URL, files=files, timeout=5)
        response.raise_for_status()

        # 5. httpbin의 응답을 통해 서버가 파일을 어떻게 받았는지 확인합니다.
        result = response.json()
        print("--- 서버가 받은 파일 정보 ---")
        print(result['files'])

except requests.exceptions.RequestException as e:
    print(f"파일 업로드에 실패했습니다: {e}")
````

**문제 6: 특정 단어가 포함된 뉴스 검색하기 (네이버 API)**

