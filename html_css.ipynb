## 📜 1단계: HTML이 뭔가요? (feat. 샌드위치)

**HTML**은 **H**yper**T**ext **M**arkup **L**anguage의 약자로, 웹 브라우저(크롬, 엣지 등)에게 "이 페이지를 어떻게 그려줘!"라고 알려주는 '설명서' 언어입니다.

HTML은 '**태그(Tag)**'라는 꺾쇠괄호(`< >`)로 모든 것을 표현해요.

마치 샌드위치처럼, '여는 태그'와 '닫는 태그'가 한 쌍을 이룹니다.

* `<p>`: 문단(Paragraph) **시작!**
* `</p>`: 문단 **끝!** (슬래시 `/`가 붙어요)

그래서 ` <p>안녕하세요</p> ` 라고 쓰면, 브라우저는 "아! '안녕하세요'는 하나의 문단이구나!"라고 알아듣습니다.

* `<h1>`: 가장 큰 제목 (Heading 1)
* `<em>`: 강조 (Emphasis, *기울임*)
* `<strong>`: 중요 (Strong, **굵게**)

우리는 파이썬 개발자니까, `index.html` 파일을 손으로 만들기보단, 파이썬으로 이 HTML '문자열'을 생성해볼까요?

````python
# 파이썬 변수에 HTML 태그가 담긴 문자열을 만듭니다.
my_title = "Hello, World Wide Web!"
my_text = "I'm a Python programmer and a bug collector."

# f-string을 사용해 HTML 조합
html_content = f"<h1>{my_title}</h1>\n<p>{my_text}</p>"

print("--- 파이썬이 생성한 HTML 코드 ---")
print(html_content)

# index.html로 저장하기
file_path = "index.html"
with open(file_path, "w", encoding="utf-8") as f:
    f.write(html_content)

print(f"\n✅ '{file_path}' 파일이 생성되었습니다. 브라우저로 열면 웹페이지가 표시됩니다.")
````

## 🏗️ 2단계: HTML의 '필수 구조' (보일러플레이트)

사실, 제대로 된 HTML 문서는 `<h1>` 태그만 덜렁 있으면 안 되고, "이 문서는 HTML입니다!"라고 알려주는 '필수 뼈대(Boilerplate)'가 필요합니다.

* `<!DOCTYPE html>`: "이건 HTML5 문서야!" (선언)
* `<html>`: "여기서부터 진짜 HTML 시작!" (모든 것을 감싸는 루트)
* `<head>`: "페이지의 '설정' 부분이야." (브라우저 탭 제목, 한글 설정, CSS 링크 등 숨은 정보)
* `<body>`: "페이지의 '실제 내용' 부분이야." (눈에 보이는 모든 것. `<h1>`, `<p>` 등)

````python
# -*- coding: utf-8 -*-
# HTML 기본 구조(보일러플레이트)를 파이썬으로 생성하고 index.html로 저장하기

page_title = "My First Python-Generated Page"
body_content = "<h2>About me</h2>\n<p>I love Python!</p>"

# HTML 템플릿 (여러 줄 문자열)
html_template = f"""<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="utf-8">
  <title>{page_title}</title>
</head>
<body>
  {body_content}
</body>
</html>
"""

print("--- 생성된 HTML 코드 ---")
print(html_template)

# index.html로 저장하기
file_path = "index.html"
with open(file_path, "w", encoding="utf-8") as f:
    f.write(html_template)

print(f"\n✅ '{file_path}' 파일이 생성되었습니다. 브라우저에서 열어보세요.")
````

## 🎨 3단계: HTML에 '옷' 입히기 (CSS 기초)

뼈대만 있으니 너무 밋밋하죠? **CSS** (Cascading Style Sheets)로 꾸며봅시다.

CSS를 추가하는 가장 쉬운 방법은 `<head>` 태그 안에 `<style>` 태그를 넣는 거예요.

CSS 문법은 간단합니다: `선택자 { 속성: 값; }`

* `body { background-color: black; }`
    * **해석**: "`<body>` 태그를 **선택**해서, `배경색` **속성**을 `검은색`**값**으로 바꿔줘!"

````python
# -*- coding: utf-8 -*-
# 3단계: CSS를 포함한 HTML 문서 생성 후 index.html로 저장하기

page_title = "My Styled Page"

# 1. CSS 스타일 정의
my_style = """
  body {
    background-color: rgb(20, 40, 60); /* 어두운 파란색 배경 */
    color: rgb(240, 248, 255);        /* 밝은 회색 텍스트 */
    font-family: Arial, sans-serif;
    text-align: center;
  }
  h1 {
    color: rgb(255, 111, 111);        /* 제목은 연어색 */
  }
  p {
    font-size: 18px;
  }
"""

# 2. HTML 본문 내용
body_content = """
  <h1>Styled by Python!</h1>
  <p>This page has custom CSS created directly from Python 🐍</p>
"""

# 3. HTML 템플릿 (head 안에 <style> 포함)
html_template = f"""<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="utf-8">
  <title>{page_title}</title>
  <style>
    {my_style}
  </style>
</head>
<body>
  {body_content}
</body>
</html>
"""

print("--- 생성된 HTML 코드 ---")
print(html_template)

# 4. index.html 파일로 저장
file_path = "index.html"
with open(file_path, "w", encoding="utf-8") as f:
    f.write(html_template)

print(f"\n✅ '{file_path}' 파일이 생성되었습니다. 브라우저에서 열면 스타일이 적용된 페이지가 표시됩니다.")
````

## ✨ 4단계: '콕' 집어서 꾸미기 (CSS Class)

`p { ... }` 라고 하면 *모든* `<p>` 태그가 바뀌어버려요.
내가 원하는 것'만' 콕 집어서 꾸미고 싶을 땐 **'class (클래스)'**를 씁니다.

1.  **HTML**: `<p class="highlight">` 처럼 '이름표'를 붙여줍니다.
2.  **CSS**: `.highlight { ... }` 처럼 '`.` (점)'을 찍어서 그 이름표를 부릅니다.

````python
# -*- coding: utf-8 -*-
# 4단계: CSS 클래스(class)를 이용해 선택적으로 꾸미기

page_title = "CSS Class Test"

# 1. CSS 스타일 정의
my_style = """
  body {
    background-color: #222;
    color: #EEE;
    font-family: Arial, sans-serif;
    text-align: center;
  }
  /* '.highlight' 클래스는 노란색 글자 + 굵게 */
  .highlight {
    color: yellow;
    font-weight: bold;
  }
  /* '.normal' 클래스는 회색 글자 */
  .normal {
    color: lightgray;
  }
"""

# 2. HTML 본문 (class 속성으로 구분)
body_content = """
  <h1>Class Test</h1>
  <p class="normal">This is normal text.</p>
  <p class="highlight">This is highlighted text!</p>
  <p class="normal">This is normal text again.</p>
"""

# 3. HTML 템플릿에 스타일 포함
html_template = f"""<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="utf-8">
  <title>{page_title}</title>
  <style>
    {my_style}
  </style>
</head>
<body>
  {body_content}
</body>
</html>
"""

print("--- 생성된 HTML 코드 ---")
print(html_template)

# 4. index.html 파일로 저장
file_path = "index.html"
with open(file_path, "w", encoding="utf-8") as f:
    f.write(html_template)

print(f"\n✅ '{file_path}' 파일이 생성되었습니다. 브라우저로 열면 'highlight' 문장만 노란색으로 표시됩니다.")
````

## 🧾 5단계: Python으로 HTML '자동 생성' (테이블 만들기)

파이썬의 진가는 '반복 작업'을 자동화할 때 나타나죠. HTML도 마찬가지입니다!

예를 들어, 100개의 이모지 목록을 HTML 테이블(`<table>`)로 만든다고 상상해보세요.
손으로 `<tr><td>...</td></tr>`를 100번 복사/붙여넣기 할 건가요? 😱

파이썬의 **`for`문**을 사용하면 단 3줄이면 끝납니다!

(`unicodedata`는 파이썬에 내장된 모듈로, 이모지의 공식 이름을 알려줍니다.)

````python
# -*- coding: utf-8 -*-
# 5단계: Python for문으로 HTML 테이블 생성 후 index.html로 저장하기
import unicodedata

# 1. 데이터 준비 (파이썬 문자열)
all_emoji = "🐞🐍👍🎉🤩😂🐶🍿😎✨"
columns = ["#", "Emoji", "Name"]

# 2. 테이블 헤더(<thead>) 만들기
table_head = "<thead>\n  <tr>"
for col in columns:
    table_head += f"<th>{col}</th>"
table_head += "\n  </tr>\n</thead>"

# 3. 테이블 바디(<tbody>) 만들기 (for문)
table_body = "<tbody>\n"
for i, emoji in enumerate(all_emoji, start=1):
    name = unicodedata.name(emoji).title()
    table_body += f"  <tr><td>{i}</td><td>{emoji}</td><td>{name}</td></tr>\n"
table_body += "</tbody>"

# 4. 전체 테이블 합치기
full_table = f"<table border='1' cellpadding='8' cellspacing='0'>\n{table_head}{table_body}\n</table>"

# 5. HTML 기본 구조와 CSS 추가
html_template = f"""<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="utf-8">
  <title>Emoji Table Generated by Python</title>
  <style>
    body {{
      background-color: #202020;
      color: #EEE;
      font-family: Arial, sans-serif;
      text-align: center;
    }}
    table {{
      margin: 20px auto;
      border-collapse: collapse;
      background-color: #333;
    }}
    th {{
      background-color: #444;
      color: yellow;
    }}
    td {{
      padding: 8px 12px;
    }}
  </style>
</head>
<body>
  <h1>🐍 Emoji Table Created by Python</h1>
  {full_table}
</body>
</html>
"""

# 6. index.html로 저장
file_path = "index.html"
with open(file_path, "w", encoding="utf-8") as f:
    f.write(html_template)

print(f"✅ '{file_path}' 파일이 생성되었습니다. 브라우저에서 열면 이모지 테이블이 표시됩니다.")
````

## 🔍 6단계: Python으로 HTML '분석'하기 (파싱)

HTML을 '만드는(Write)' 것을 배웠으니, '읽는(Parse)' 것도 배워봐야겠죠?

**파싱(Parsing)**은 HTML 코드 덩어리에서 내가 원하는 정보(예: 모든 이미지 주소, 모든 링크)만 '추출'하는 기술입니다.

파이썬은 `html.parser`라는 강력한 기본 라이브러리를 제공합니다. (물론 `BeautifulSoup`같은 외부 라이브러리가 더 유명하긴 해요!)

`HTMLParser`를 사용하려면, 이 클래스를 **'상속'**받아서 `handle_starttag` 라는 메서드를 '오버라이딩(재정의)'하면 됩니다.

````python
from html.parser import HTMLParser

# 1. HTMLParser를 '상속'받아 나만의 파서(ImageParser)를 만듭니다.
class ImageParser(HTMLParser):

  # 2. 'handle_starttag' 메서드는 여는 태그(<...>)를 만날 때마다 자동 호출됩니다.
  def handle_starttag(self, tag, attrs):
    # 'tag'는 태그 이름 (예: 'h1', 'p', 'img')
    # 'attrs'는 속성 리스트 (예: [('src', 'sky.png'), ('alt', 'A sky')])

    # 3. 만약 태그가 'img' 라면?
    if tag == "img":
      # 4. 속성 리스트(attrs)를 돌면서 'src'를 찾습니다.
      for attr_name, attr_value in attrs:
        if attr_name == "src":
          print(f"[ImageParser] 발견! 이미지 주소: {attr_value}")

# --- 테스트 해보기 ---

# 5. 분석할 HTML 데이터를 준비합니다.
html_data = """
<h1>My Gallery</h1>
<p>Here are my photos:</p>
<img src="sky_1.png" alt="Cloudy sky.">
<div>
  <img src="images/gallery/sky_2.png" alt="Sun.">
</div>
"""

# 6. 파서를 생성하고, 'feed' 메서드로 HTML 데이터를 '먹여줍니다'.
parser = ImageParser()
parser.feed(html_data)
````

### 문제 1: `for`문으로 HTML 리스트(`<ul>`) 만들기

**미션**: `my_list = ["Python", "Flask", "Django"]` 리스트를 사용해, HTML '순서 없는 목록'(`<ul>`)을 만드세요.

### 문제 2: 딕셔너리로 HTML 링크(`<a>`) 목록 만들기

**미션**: `links = {"Real Python": "https://realpython.com", "Python.org": "https://www.python.org"}` 딕셔너리를 사용해 `<a>` 태그 목록을 만드세요.
**힌트**: `for key, value in links.items():`를 사용하세요. 링크 주소는 `href` 속성에 들어갑니다.

````python
links = {
    "Real Python": "https://realpython.com",
    "Python.org": "https://www.python.org"
}

html_links = ""
for text, url in links.items():
  # <a href="주소">텍스트</a> 형태
  html_links += f'<a href="{url}">{text}</a><br>\n' # <br>은 줄바꿈 태그

print(html_links)
````

### 문제 3: `if`문으로 CSS `class` 동적 적용하기

**미션**: `scores = [95, 80, 45, 100]` 리스트가 있습니다. 90점 이상은 `pass` 클래스, 미만은 `fail` 클래스를 적용하여 `<p>` 태그 리스트를 만드세요.
**힌트**: `for`문 안에서 `if` 문을 사용해 `class_name` 변수를 다르게 설정하세요.

````python
scores = [95, 80, 45, 100]

# 미리 스타일을 정의해둡니다.
style = "<style> .pass { color: green; } .fail { color: red; } </style>\n"
body = ""

for score in scores:
  if score >= 90:
    class_name = "pass"
  else:
    class_name = "fail"

  # f-string 안에서 변수를 사용해 class 이름 적용
  body += f'<p class="{class_name}">Score: {score}</p>\n'

print(style + body)
````

```
Score: 100 (녹색 글자)
```

### 문제 4: `HTMLParser`로 모든 헤드라인(`<h1>`, `<h2>`) 텍스트 추출

**미션**: `handle_starttag`에서 `h1` 또는 `h2` 태그를 만나면 '플래그'를 켜고, `handle_data`에서 플래그가 켜진 상태일 때만 텍스트를 출력하세요.
**힌트**: `__init__`에서 `self.is_heading = False`로 플래그를 만드세요.

````python
from html.parser import HTMLParser

class HeadingParser(HTMLParser):
  def __init__(self):
    super().__init__()
    self.is_heading = False  # 플래그(flag) 변수

  def handle_starttag(self, tag, attrs):
    if tag in ["h1", "h2"]:  # 'h1' 또는 'h2' 태그가 시작되면
      self.is_heading = True

  def handle_endtag(self, tag):
    if tag in ["h1", "h2"]:  # 태그가 닫히면
      self.is_heading = False

  def handle_data(self, data):
    if self.is_heading:  # 플래그가 True일 때만
      print(f"발견된 헤드라인: {data.strip()}")

html_data = "<h1>Main Title</h1><p>Some text.</p><h2>Subtitle</h2><p>More text.</p>"

parser = HeadingParser()
parser.feed(html_data)
````

### 문제 5: `enumerate`로 `<img>` 갤러리 `alt` 속성 만들기

**미션**: `image_files = ["img1.png", "img2.jpg", "logo.svg"]` 리스트로 `<img>` 태그들을 만드세요.
**힌트**: `alt` 속성은 접근성을 위해 중요합니다. `enumerate`를 사용해 "Image 1", "Image 2"... 처럼 동적인 `alt` 텍스트를 만드세요.

````python
# 문제 5: enumerate로 <img> 갤러리 alt 속성 만들기

image_files = ["img1.png", "img2.jpg", "logo.svg"]

gallery_html = ""

for i, img_file in enumerate(image_files, start=1):  # 1부터 번호 세기
    alt_text = f"Gallery Image {i}"                 # 동적인 alt 텍스트
    gallery_html += f'<img src="{img_file}" alt="{alt_text}">\n'

print(gallery_html)
````

### 문제 6: `html.entities`로 특수 문자 변환하기

**미션**: '보다 작다' 기호 `<` 는 HTML 태그로 오해받을 수 있습니다. 파이썬 `html` 모듈을 사용해 이 기호의 HTML 엔티티 이름(예: `lt`)을 찾아내세요.
**힌트**: `ord('<')`로 코드 포인트를 찾고, `html.entities.codepoint2name` 딕셔너리를 사용하세요.

````python
import html

char = "<"
code_point = ord(char)  # 문자를 코드 포인트(숫자)로 변환

# 코드 포인트에 해당하는 엔티티 이름을 딕셔너리에서 찾기
entity_name = html.entities.codepoint2name[code_point]

print(f"'{char}'의 HTML 엔티티 이름: {entity_name}")
print(f"HTML에서 사용법: &{entity_name};")
````

**힌트**: (고급) `users[0].keys()`로 헤더(`<th>`)를 먼저 만들고, `for user in users:` 루프에서 `user.values()`를 사용해 `<td>`를 만드세요.

### 문제 10: `HTMLParser`로 `alt` 텍스트가 비어있는 `<img>` 태그 찾기

**미션**: 웹 접근성을 위해 `alt` 텍스트는 매우 중요합니다. `HTMLParser`를 사용해 `alt` 속성이 비어있거나(`alt=""`) 없는 `<img>` 태그의 `src`를 출력하세요.
**힌트**: `attrs`를 `dict(attrs)`로 변환하면 `alt = attr_dict.get("alt", "")` 처럼 쉽게 값을 가져올 수 있습니다.

````python
class AltTextParser(HTMLParser):
  def handle_starttag(self, tag, attrs):
    if tag == "img":
      # attrs 리스트를 딕셔너리로 바꾸면 작업이 편합니다.
      attr_dict = dict(attrs)

      src = attr_dict.get("src", "(주소 없음)")
      # 'alt' 키를 가져오되, 없으면 기본값으로 빈 문자열("")을 사용
      alt = attr_dict.get("___", "")

      if not alt: # alt 텍스트가 비어있다면 (if alt == "":)
        print(f"경고! [접근성 오류] alt 텍스트가 없습니다: {src}")

html_data = """
<img src="a.png" alt="A cute cat.">
<img src="b.png" alt="">
<img src="c.png">
"""
parser = AltTextParser()
parser.feed(html_data)
````
