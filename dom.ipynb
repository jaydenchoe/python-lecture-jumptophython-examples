# 🤖 DOM 개념을 활용한 나만의 파이썬 챗봇 만들기

안녕하세요! 오늘 우리는 **DOM(문서 객체 모델)**이라는 개념을 살짝 맛보고, 이 아이디어를 활용해서 아주 간단한 파이썬 챗봇을 만들어 볼 거예요.

**DOM이 뭔가요?**

DOM은 'Document Object Model'의 약자예요. 웹페이지(HTML 문서)를 '객체'의 '모델', 즉 **논리적인 트리 구조**로 표현하는 방식이에요. 

마치 가족 관계도처럼, 웹페이지 전체(문서)가 가장 큰 할아버지(Document)고, 그 아래에 자식들(HTML), 또 그 아래에 자식들(HEAD, BODY)이 있는 것처럼 계층을 이뤄요. 

이렇게 구조화하면 프로그래밍 언어(주로 JavaScript)가 웹페이지의 특정 부분(예: '저기 세 번째 문단')을 쉽게 찾아가서 내용을 바꾸거나(예: '텍스트를 이걸로 바꿔!') 스타일을 변경할 수 있게 됩니다.

**우리의 챗봇과 무슨 상관이죠?**

우리는 웹페이지 대신, 챗봇의 '지식'을 구조화할 거예요. DOM이 웹 문서를 객체로 다루듯, 우리는 챗봇의 지식을 파이썬의 **딕셔너리(Dictionary)**라는 자료 구조로 만들 거예요. 딕셔너리는 '키(Key)'와 '값(Value)'으로 데이터를 묶어서 관리하는데, 이게 바로 챗봇의 '질문'과 '대답'을 짝지어 저장하는 멋진 '지식 모델'이 될 수 있답니다!

자, 그럼 파이썬의 가장 기초적인 '말하기'부터 시작해볼까요?

## 1단계: 챗봇의 첫 마디 (print 사용하기)

파이썬에서 화면에 글자를 출력하는 가장 기본적인 명령어는 `print()`입니다. 괄호 안에 따옴표(`"` 또는 `'`)로 묶은 글자를 넣으면, 챗봇이 그 말을 하게 됩니다.

```python
# 'Shift + Enter'를 눌러 이 코드 셀을 실행해보세요.
print("안녕하세요! 저는 파이썬 챗봇 '파이-챗'입니다.")
print("만나서 반가워요!")
```

## 2단계: 챗봇의 이름 저장하기 (변수 사용하기)

DOM에서 각 노드(객체)가 자신의 정보(예: 텍스트 내용)를 가지는 것처럼, 우리도 데이터를 '변수(Variable)'라는 상자에 저장할 수 있습니다. 변수에 이름을 붙여주면 나중에 그 이름을 불러서 상자 안의 내용물을 쉽게 꺼내 쓸 수 있어요.

`bot_name = "파이-챗"` 처럼요.

```python
# 챗봇의 이름을 'bot_name'이라는 변수에 저장합니다.
bot_name = "파이-챗"

# f-string을 사용하면 변수의 내용을 글자와 쉽게 합칠 수 있어요.
print(f"제 이름은 {bot_name}이에요.")
print(f"{bot_name}은 무엇이든 도와드릴 준비가 되어있어요!")
```

## 3단계: 사용자에게 말 걸기 (input 사용하기)

챗봇은 일방적으로 말만 하면 안 되겠죠? 사용자의 입력을 받아야 해요. `input()` 함수는 사용자로부터 키보드 입력을 받아 그 내용을 '변수'에 저장해줍니다. 

이것은 DOM에서 '클릭 이벤트'처럼 사용자의 행동에 반응하는 것과 비슷해요!

```python
# input() 안의 글자는 사용자에게 보여줄 질문입니다.
user_name = input("당신의 이름은 무엇인가요? ")

# 사용자가 입력한 이름을 {user_name} 변수로 받아서 인사합니다.
print(f"와, {user_name}님! 정말 멋진 이름이네요.")
print(f"저는 {bot_name}입니다. 반가워요, {user_name}님!")
```

## 4단계: 챗봇의 '지식 모델' 만들기 (딕셔너리)

드디어 챗봇의 '뇌'를 만들 차례입니다. DOM이 문서를 구조화하듯, 우리는 '질문(Key)'과 '대답(Value)'을 짝지어 딕셔너리(Dictionary)로 만들 거예요.

딕셔너리는 중괄호 `{}`를 사용하고, `"키": "값"` 형태로 짝을 이룹니다. 각 짝은 쉼표(`,`)로 구분해요.

```python
# 챗봇의 지식창고 (Knowledge Base)를 딕셔너리로 만듭니다.
knowledge_base = {
    "안녕": "안녕하세요! 만나서 반가워요.",
    "이름이 뭐야?": f"제 이름은 {bot_name}입니다.",
    "날씨 어때?": "오늘은 코딩하기 딱 좋은 맑은 날씨네요!",
    "취미가 뭐야?": "저는 {user_name}님과 대화하는 게 취미랍니다."
}

# 딕셔너리에서 '키'를 이용해 '값'을 꺼내볼까요?
question = "안녕"
print(f"질문: {question}")
print(f"대답: {knowledge_base[question]}")

question = "날씨 어때?"
print(f"질문: {question}")
print(f"대답: {knowledge_base[question]}")
```

## 5단계: 묻고 답하기 (if 조건문)

이제 사용자의 입력에 따라 챗봇이 '스스로' 지식창고에서 답을 찾아 말하게 해봅시다. 

`if`, `elif` (else if), `else` 조건문을 사용하면 "만약 사용자가 '안녕'이라고 말하면, 이렇게 대답해!", "그게 아니고 '날씨'라고 말하면, 저렇게 대답해!", "그 외의 말을 하면, '모르겠다'고 대답해!" 처럼 논리적인 흐름을 만들 수 있습니다.

```python
# 사용자의 질문을 받습니다.
user_input = input("무엇이 궁금한가요? (안녕, 이름이 뭐야?, 날씨 어때?) ")

# 사용자의 입력(user_input)이 딕셔너리의 '키' 중에 있는지 확인합니다.
if user_input in knowledge_base:
    # 만약 키가 존재한다면?
    print(f"파이-챗: {knowledge_base[user_input]}")
else:
    # 만약 키가 존재하지 않는다면?
    print("파이-챗: 앗, 그 말은 아직 제가 배우지 못했어요. 😥")
```

## 6단계: 대화 계속하기 (while 반복문)

지금 챗봇은 한 번 대답하고 끝나버리죠? 사용자가 '종료'라고 말할 때까지 계속 대화를 이어가도록 `while True:` 무한 반복문을 사용해봅시다.

반복문 안에서 '종료'라는 특정 입력이 들어오면, `break` 명령어로 반복을 멈추고 챗봇을 종료합니다.

```python
print("챗봇을 시작합니다. 대화를 끝내고 싶으면 '종료'라고 입력하세요.")

while True:  # 무한 반복 시작!
    user_input = input("당신: ")

    if user_input == "종료":
        print("파이-챗: 안녕히 가세요! 다음에 또 만나요.")
        break  # 반복문을 탈출합니다.

    elif user_input in knowledge_base:
        print(f"파이-챗: {knowledge_base[user_input]}")
        
    else:
        print("파이-챗: 앗, 그 말은 아직 제가 배우지 못했어요. 😥")
```

## 7단계 (심화): 챗봇의 '지식' 수정하기 (딕셔너리 값 변경)

DOM의 가장 강력한 기능은 문서를 '수정'하는 것입니다. 우리 챗봇도 대화 중에 새로운 지식을 배울 수 있게 해볼까요?

모르는 말이 들어왔을 때(`else` 부분), 사용자에게 되물어보고 딕셔너리에 새로운 '키'와 '값'을 추가하는 겁니다. 딕셔너리에 값을 추가하는 건 `딕셔너리이름[새로운키] = 새로운값` 처럼 간단해요!

```python
print("챗봇을 시작합니다. 대화를 끝내고 싶으면 '종료'라고 입력하세요.")
print("--- (학습 기능 탑재!) ---")

while True:
    user_input = input("당신: ")

    if user_input == "종료":
        print("파이-챗: 안녕히 가세요! 많이 배울 수 있어서 즐거웠어요.")
        break

    elif user_input in knowledge_base:
        print(f"파이-챗: {knowledge_base[user_input]}")
        
    else:
        print(f"파이-챗: 앗, '{user_input}'은(는) 처음 듣는 말이에요.")
        new_answer = input(f"제가 뭐라고 대답하면 좋을까요? ")
        
        # 챗봇의 지식(DOM)을 수정합니다!
        knowledge_base[user_input] = new_answer
        
        print("파이-챗: 와! 새로운 걸 배웠어요. 고마워요!")
```
